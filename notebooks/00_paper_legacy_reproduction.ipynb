# Paper legacy reproduction

This notebook renders the destination-only, top-k legacy result-generation path. It is a regression view, not the corrected baseline.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

dataset_id = '202604081400'
run_name = 'paper_legacy'
run_dir = Path('results') / dataset_id / run_name
manifest = json.loads((run_dir / 'run_manifest.json').read_text())
prefixes = pd.read_csv(run_dir / 'prefixes.csv')
print({'dataset': dataset_id, 'run': run_name, 'config_hash': manifest.get('config', {}).get('hash'), 'input_sha256': manifest.get('input', {}).get('sha256'), 'git_commit': manifest.get('git_commit')})
```

```python
displayed = prefixes.loc[(prefixes['selected']) & (prefixes['flow_count'] >= 1000)].sort_values('flow_count', ascending=False).head(10).copy()
display(displayed[['normalized_dst_prefix', 'flow_count', 'packet_count', 'byte_count', 'score']])
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for axis, column, title in zip(axes, ['packet_count', 'byte_count', 'flow_count'], ['Packets', 'Frame bytes', 'Flows']):
    axis.bar(displayed['normalized_dst_prefix'], displayed[column])
    axis.set_title(title)
    axis.tick_params(axis='x', rotation=90)
fig.tight_layout()
```
